Ejercicio 2

```python
df2=pd.read_csv("cuentas_credicel.csv", encoding="latin")
df2
```

```python
data=df2.copy()
```

```python
fechas=["fecha_ultimo_pago", "fecha_pago_proximo","fecha"]
for col in fechas:
    data[col] = data[col].fillna(pd.to_datetime("01/01/2030", format="%d/%m/%Y"))
```

```python
columnas_objetos = data.select_dtypes(include=['object']).columns
for col in columnas_objetos:
    data[col] = data[col].fillna("Sin registro")
```

```python
columnas_numericos=data.select_dtypes(include=['int64', 'float64']).columns
for col in columnas_numericos:
    data[col] = data[col].fillna("Promedio")
    
valores_nulos=data.isnull().sum()
valores_nulos
```

```python
data.to_csv("Valores_Nulos_OmarQuinn.csv")
```

```python
dataframe = pd.read_csv("Valores_Nulos_OmarQuinn.csv", encoding= "latin")
```

```python
dataframe.info()
```

```python
cuantitativas = dataframe.iloc[:, [0]+[1]+[3]+list(range(8,15))+[17]+[18]+[20]+[21]+[22]+[29]+[30]]
cualitativas = dataframe.iloc[:, list(range(4,8))+[15]+[16]+[19]+list(range(23,29))+list(range(31,38))]
```

```python
y=cuantitativas
Limite_Superior= y.mean() + 3*y.std()
Limite_Inferior= y.mean() - 3*y.std()
print("Limite superior permitido", Limite_Superior)
print("Limite inferior permitido", Limite_Inferior)
```

```python
data3= cuantitativas[(y<=Limite_Superior)&(y>=Limite_Inferior)]
data3
```

```python
valores_nulos=data3.isnull().sum()
valores_nulos
```

```python
data_clean=data3.copy()
data_clean=data_clean.fillna(round(data3.mean(),1))

valores_nulos=data_clean.isnull().sum()
valores_nulos
```

```python
y=cuantitativas

percentile25=y.quantile(0.25) 
percentile75=y.quantile(0.75) 
iqr= percentile75 - percentile25

Limite_Superior_iqr= percentile75 + 1.5*iqr
Limite_Inferior_iqr= percentile25 - 1.5*iqr
print("Limite superior permitido", Limite_Superior_iqr)
print("Limite inferior permitido", Limite_Inferior_iqr)
```

```python
data3_iqr= cuantitativas[(y<=Limite_Superior_iqr)&(y>=Limite_Inferior_iqr)]
data3_iqr
```

```python
data4_iqr=data3_iqr.copy()
data4_iqr=data4_iqr.fillna(round(data3_iqr.median(),1))
valores_nulos=data3_iqr.isnull().sum()
valores_nulos
```

```python
Datos_limpios = pd.concat([cualitativas, data4_iqr], axis=1)
valores_nulos=Datos_limpios.isnull().sum()
valores_nulos
```

```python
Datos_limpios.to_csv("Ejercicio3_OmarQuinn.csv")
```
