```python
import libreria as f
import numpy as np
import pandas as pd 
```

```python
df= f.cargar_csv("Budapest_Hungria.csv")
```

```python
df = pd.DataFrame({
    "pais":     ["MX",   None,    "AR",     None],
    "ciudad":   ["CDMX", "Puebla", None,    "Monterrey"],
    "precio":   [100.0,  np.nan,   120.0,   np.nan],
    "cantidad": [10,     12,       np.nan,  15],
    "fecha":    pd.to_datetime(["2024-01-01", None, "2024-01-03", "2024-01-04"]),
    "serie":    [1,      np.nan,   np.nan,  4]
})
```

```python
print("Nulos iniciales por columna:\n", df.isna().sum(), "\n")
```

```python
# 1) Texto -> cadena
df_txt = f.metodo_string(df, "DESCONOCIDO")
print("Texto rellenado:\n", df_txt[["pais","ciudad"]], "\n")
```

```python
# 2) Numéricas -> promedio por columna
df_mean = f.promedio(df)
print("Promedio (precio/cantidad/serie):\n", df_mean[["precio","cantidad","serie"]], "\n")
```

```python
# 3) Numéricas -> mediana por columna
df_median = f.mediana(df)
print("Mediana (precio/cantidad/serie):\n", df_median[["precio","cantidad","serie"]], "\n")
```

```python
# 4) Numéricas -> constante
df_const = f.constante(df, 0)
print("Constante=0 (precio/cantidad/serie):\n", df_const[["precio","cantidad","serie"]], "\n")
```

```python
# 5) Forward fill (elige columnas numéricas)
cols_num = ["precio", "cantidad", "serie"]
df_ff = f.forward_fill(df)
print("Forward fill:\n", df_ff, "\n")
```

```python
df_bf = f.backward_fill(df)
print("Backward fill:\n", df_bf, "\n")
```

```python
print("Resumen de nulos del DF:\n", f.nulos_en_dataframe(df))
```

```python
df_filas=f.filas_impares(df)
print("filas impares: \n", df_filas, "\n")
```
