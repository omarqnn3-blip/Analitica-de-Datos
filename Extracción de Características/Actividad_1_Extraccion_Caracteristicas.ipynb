```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
TaxonName = Inventories["TaxonName"].value_counts().reset_index()
TaxonName
```

```python
filtroTaxonName = TaxonName[TaxonName['count']>20500]
filtroTaxonName
```

```python
filtro_indexTaxonName = filtroTaxonName.set_index('TaxonName')
filtro_indexTaxonName
```

```python
filtro_indexTaxonName.plot(kind='bar', width=0.8, figsize=(8,3), color= "red")
plt.title('TaxonName')
plt.xlabel('Tipo de Organismo')
plt.ylabel('Frecuencia')
```

```python
filtro_indexTaxonName.plot(kind='area', figsize=(20,4),alpha = 0.5)
```

```python
filtro_indexTaxonName["count"].plot(kind='pie', figsize=(20,10), shadow=True, autopct="%0.1f %%")
```
